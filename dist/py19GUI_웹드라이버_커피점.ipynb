```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.by import By          
from selenium.webdriver.common.keys import Keys        
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup as bs
import pandas as pd
import time as tt

# 1. 크롬 옵션 설정
options = Options()
options.add_argument("--window-size=1920,1080") 
# options.add_argument("--no-sandbox")
# options.add_argument("--disable-dev-shm-usage")
# options.add_argument("--remote-allow-origins=*")

# 💡 Selenium 4.6+ 의 Selenium Manager가 Chrome 버전에 맞는 드라이버를 자동으로 찾아 다운로드합니다.
driver = webdriver.Chrome(options=options)

# 클릭을 편리하게 해주는 gogo 함수 정의
def gogo(sel):
    driver.find_element(By.CSS_SELECTOR, sel).click()

# 3. 사이트 접속
driver.get('https://tenpercentcoffee.com/')
print("페이지 제목:", driver.title)
tt.sleep(2) 

# 4. 마우스 호버할 요소 찾기 및 실행
hover_element = driver.find_element(By.CSS_SELECTOR, "#header > div")
action = ActionChains(driver)
action.move_to_element(hover_element).perform()
print('메뉴 여는중...')
tt.sleep(1)

# 5. 매장 찾기 메뉴 클릭
gogo("#header > div > div > ul > li:nth-child(4) > ul > li > a")
print('매장 위치 검색 페이지 이동 중...')
tt.sleep(3) 

print("매장 위치 로드 완료")

# 6. 검색창 클릭 후 '부산광역시' 입력 및 엔터
gogo("#q")
tt.sleep(1)
driver.find_element(By.CSS_SELECTOR, "#q").send_keys("서울특별시")
driver.find_element(By.CSS_SELECTOR, "#q").send_keys(Keys.ENTER) 
print('서울특별시 검색어 입력 완료')

# 💡 비동기 데이터가 로드될 때까지 최대 10초간 대기
try:
    WebDriverWait(driver, 10).until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "#area_list > li"))
    )
    print("매장 검색 결과 로딩 완료!")
except:
    print("로딩 시간이 초과되었거나 검색 결과가 없습니다.")

# 7. BeautifulSoup으로 데이터 파싱
html = driver.page_source
soup = bs(html, 'html.parser')

# 매장 리스트 요소 선택
contents = soup.select('#area_list > li')

temp = []
print(f"찾은 매장 개수: {len(contents)}개")

for c in contents:
    title_element = c.select_one('strong') 
    
    if title_element:
        store_name = title_element.text.strip()
        print(f'성공! 매장명: {store_name}')
        temp.append(store_name)
    else:
        store_info = c.text.strip().replace('\n', ' ')
        print(f'성공! 정보: {store_info}')
        temp.append(store_info)

# 결과 출력
print("\n--- 수집된 매장 리스트 ---")
print(temp)

# 8. 브라우저 종료
# driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup as bs
import pandas as pd
import requests
import time as tt
import re

options = Options()
options.add_argument("--window-size=1920,1080")
driver = webdriver.Chrome(options=options)
wait = WebDriverWait(driver, 15)

def js_click(sel):
    el = driver.find_element(By.CSS_SELECTOR, sel)
    driver.execute_script("arguments[0].click();", el)

# ── 1. 접속 → 매장찾기 페이지 이동 ──────────────────────
driver.get('https://tenpercentcoffee.com/')
tt.sleep(2)

hover_element = driver.find_element(By.CSS_SELECTOR, "#header > div")
ActionChains(driver).move_to_element(hover_element).perform()
tt.sleep(1.5)

MENU_SEL = "#header > div > div > ul > li:nth-child(4) > ul > li > a"
try:
    wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, MENU_SEL))).click()
    print('✅ 일반 클릭 성공')
except:
    js_click(MENU_SEL)
    print('✅ JS 강제 클릭 성공')
tt.sleep(3)

# ── 2. Selenium 쿠키 → requests 세션 복사 ───────────────
session = requests.Session()
for cookie in driver.get_cookies():
    session.cookies.set(cookie['name'], cookie['value'])

headers = {
    'Referer'         : 'https://tenpercentcoffee.com/sub/sub41.php',
    'X-Requested-With': 'XMLHttpRequest',
    'User-Agent'      : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
                        'AppleWebKit/537.36 (KHTML, like Gecko) '
                        'Chrome/120.0.0.0 Safari/537.36',
}

# ── 3. API 호출 ──────────────────────────────────────────
API_URL     = 'https://tenpercentcoffee.com/sub/_inc.ajax.arealist.php'
SEARCH_WORD = '서울특별시'

response = session.get(API_URL, params={'q': SEARCH_WORD},
                       headers=headers, timeout=15)
print(f'✅ API 응답: {response.status_code} / {len(response.text)}글자')

# ── 4. 🔑 핵심 수정: div.rlwrap 으로 매장 1개씩 선택 ────
soup  = bs(response.text, 'html.parser')

# 💡 구조:
# div.rlwcal (전체 컨테이너 1개)
#   └─ a[onclick] (매장 1개)
#        └─ div.rlwrap (매장 카드)  ← 이걸 선택!
#             ├─ p.rlwbp  : 매장명
#             ├─ p.rlwsp  : 주소
#             └─ div.new_event : 이벤트
store_cards = soup.select('div.rlwrap')
print(f'🏪 총 매장 수: {len(store_cards)}개')

# ── 5. 데이터 추출 ────────────────────────────────────────
temp = []
for card in store_cards:

    # 매장명
    name_el = card.select_one('p.rlwbp')
    name    = name_el.get_text(strip=True) if name_el else '이름없음'

    # 주소
    addr_el = card.select_one('p.rlwsp')
    addr    = addr_el.get_text(strip=True) if addr_el else '주소없음'

    # 이벤트 (없는 매장은 '-')
    event_el = card.select_one('div.new_event')
    event    = event_el.get_text(strip=True) if event_el else '-'

    # 위도/경도/매장ID → 부모 <a> 태그의 onclick에서 추출
    # setMove(위도, 경도, '매장ID', 줌레벨)
    a_tag = card.find_parent('a')   # 💡 부모 a 태그 찾기
    lat, lng, store_id = '', '', ''
    if a_tag:
        onclick = a_tag.get('onclick', '')
        # 정규식: 숫자(소수점 포함) 패턴 추출
        nums = re.findall(r'[\d.]+', onclick)
        if len(nums) >= 3:
            lat      = nums[0]   # 위도
            lng      = nums[1]   # 경도
            store_id = nums[2]   # 매장 ID

    temp.append({
        '매장명' : name,
        '주소'   : addr,
        '이벤트' : event,
        '위도'   : lat,
        '경도'   : lng,
        '매장ID' : store_id
    })

# ── 6. 결과 출력 & CSV 저장 ──────────────────────────────
df = pd.DataFrame(temp)
print(f'\n--- 최종 수집 결과 ({len(df)}개) ---')
print(df.to_string())

df.to_csv('텐퍼센트_서울매장.csv', index=False, encoding='utf-8-sig')
print('\n✅ CSV 저장 완료: 텐퍼센트_서울매장.csv')

driver.quit()
```
